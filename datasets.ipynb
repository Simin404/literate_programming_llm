```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Data
df = pd.DataFrame({
    "Dataset": [
        "HumanEval", "HumanEval-ET", "HumanEval-X", "MBPP", "MBPP-ET",
        "CodeNet", "PIE", "BT-transcoder", "Codesearchnet", "MCMD",
        "BigCloneBench", "Rosetta Code"
    ],
    "Total": [
        164, 164, 4920, 1000, 1000,
        14_000_000, 77_000, 8380, 6_000_000, 500,
        1140, 114_376
    ],
    "#P_Language": [
        1, 1, 6, 1, 1,
        55, 1, 16, 6, 500,
        1, 926
    ],
    "#P_Task": [
        164, 164, 164, 974, 974,
        3474, 2530, 524, 2_000_000, 5,
        1140, 1228
    ]
})

# Data points
x = df["#P_Language"].values
y = df["#P_Task"].values
z = np.log10(df["Total"].values + 1)  # Use log-scaled Total as Z
sizes = z * 100  # Bubble size

# Plot
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')

# Scatter bubbles at real Z height
ax.scatter(x, y, z, s=sizes, c=sizes, cmap='plasma', alpha=0.7)

# Add dataset name labels at same location as bubbles
for i in range(len(df)):
    ax.text(x[i], y[i], z[i] + 0.1, df["Dataset"][i],
            fontsize=9, color='black', ha='center', va='bottom')

# Axes and title
ax.set_xlabel("#P_Language")
ax.set_ylabel("#P_Task")
ax.set_zlabel("Log₁₀(Total Data Points)")
ax.set_title("3D Bubble Plot (Log Z) with Dataset Labels")

plt.tight_layout()
plt.show()
```
